```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5])
array
```

```python
adicao = np.add(array, 10)
adicao
```

```python
subtracao = np.subtract(array, 5)
subtracao
```

```python
multiplicacao = np.multiply(array, 2)
multiplicacao
```

```python
divisao = np.divide(array, 2)
divisao
```

```python
#Funções trigonométricas
seno = np.sin(array)
seno
```

```python
cosseno = np.cos(array)
cosseno
```

```python
#Funções estatísticas
media = np.mean(array)
media
```

```python
desvio_padrao = np.std(array)
desvio_padrao
```

```python
#Funções de comparação
comparacao_maior_que_3 = array > 3
comparacao_maior_que_3
```

```python
#Combinando funções
resultado = np.sqrt(array) + np.power(array, 2)
resultado
```

```python
#Funções exponenciais
exponencial = np.exp(array)
exponencial
```

```python
#Função de logaritmo
logaritmo = np.log(array)
logaritmo
```

```python
#Função de arredondamento
arredondado = np.round(array / 2)
arredondado
```

```python
#Função de potência
potencia = np.power(array, 3)
potencia
```
